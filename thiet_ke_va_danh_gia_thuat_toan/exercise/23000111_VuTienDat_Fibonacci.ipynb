```python
import time
import matplotlib.pyplot as plt

# Fibonacci đệ quy
def fib_recursive(n):
    if n <= 1:
        return n
    return fib_recursive(n-1) + fib_recursive(n-2)

# Fibonacci lặp
def fib_iterative(n):
    a, b = 0, 1
    for i in range(n):
        a, b = b, a + b
    return a

n_values = []
recursive_times = []
iterative_times = []

# thử với nhiều giá trị n
for n in range(5, 100, 10):
    n_values.append(n)

    start = time.time()
    fib_recursive(n)
    recursive_times.append(time.time() - start)

    start = time.time()
    fib_iterative(n)
    iterative_times.append(time.time() - start)

# vẽ biểu đồ
plt.plot(n_values, recursive_times, marker='o', label="Recursive")
plt.plot(n_values, iterative_times, marker='o', label="Iterative")

plt.xlabel("n")
plt.ylabel("Execution Time (seconds)")
plt.title("23000111 - Vũ Tiến Đạt ")
plt.legend()
plt.show()
```
