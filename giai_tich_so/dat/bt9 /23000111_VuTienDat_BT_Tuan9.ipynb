```python
import numpy as np
import pandas as pd
from fractions import Fraction # goi phan so sao cho dep
```

### print matrix

```python
# print matrix
def to_frac(matrix):
    rows = len(matrix)
    cols = len(matrix[0])
    return [[Fraction(matrix[i][j]).limit_denominator(1000) for j in range(cols)] for i in range(rows)]


def format_frac(f):
    if not isinstance(f, Fraction):
        f = Fraction(f).limit_denominator(1000)
    if f.denominator == 1:
        return str(f.numerator)
    return f"{f.numerator} / {f.denominator}"

def print_matrix_explore(matrix, n):
    print("Ma tran mo rong la:")
    col_width = 10
    header = " "
    for j in range(n):
        header += f" x{j + 1}".center(col_width)
    header += " | "
    header += " b".center(col_width)
    print(header)
    print("    " + "-" * (col_width * (n + 1) + 3))
    
    for i in range(len(matrix)):
        row_str = f" R{i + 1} |"
        for j in range(n):
            val = format_frac(matrix[i][j])
            row_str += str(val).center(col_width)
        row_str += " | "
        row_str += str(format_frac(matrix[i][n])).center(col_width)
        print(row_str)
    
    print("    " + "-" * (col_width * (n + 1) + 3))
    
def print_matrix(matrix, n):
    print("Ma tran:")
    col_width = 10
    header = " "
    for j in range(n):
        header += f" x{j + 1}".center(col_width)
    header += " | "
    print(header)
    print("    " + "-" * (col_width * n + 3))
    
    for i in range(len(matrix)):
        row_str = f" R{i + 1} |"
        for j in range(n):
            val = format_frac(matrix[i][j])
            row_str += str(val).center(col_width)
        row_str += " | "
        print(row_str)
    
    print("    " + "-" * (col_width * n + 3))

print("Print Completed")

```

### Phuong phap the

```python
# phuong phap the doi voi ma tran 
def iterative_method_upper_triangular_matrix(arr, y, n):
    x = [0]*n
    
    for i in range(n-1, -1, -1):
        s = y[i]
        for j in range(i+1, n):
            s -= arr[i][j]*x[j]
        
        x[i] = s / arr[i][i]
        
    print("Key")
    for i in range(n):
        val = format_frac(x[i])
        print(f"  x{i+1} = {val:>10}")
    return x

    
def iterative_method_lower_triangular_matrix(arr, b, n):
    x = [0]*n
    
    for i in range(n):
        s = b[i]
        for j in range(i):
            s -= arr[i][j]*x[j]
        
        x[i] = s
        
    print("Key")
    for i in range(n):
        val = format_frac(x[i])
        print(f"  y{i+1} = {val:>10}")
    return x
```

### Phan tich PLU 

```python
def PLU(A):
    n = len(A)
    p = np.identity(n)
    step = 0
    
    row = 0
    p_i = 0
    
    for col in range(n):
        if row >= n:
            break
        
        pivot_max = abs(A[row][col])
        pivot_swap = row
        for i in range(row + 1, n):
            if pivot_max < abs(A[i][col]):
                pivot_max = abs(A[i][col])
                pivot_swap = i
                
        if (pivot_swap != row):
            step += 1
            p_i += 1
            print(f"\n Bước {step}: R{row + 1} ↔ R{pivot_swap + 1} ")
            swap_row(A, row, pivot_swap)
            print_matrix(A, n)
            print(f"Ma tran hoan vi P{p_i} la:")
            swap_row(p, row, pivot_swap)
            print_matrix(p, n)
        
        if A[row][col] == 0:
            if (row != n - 1):
                print("Khong the phan tich A thanh ma tran LU")
            else:
                if (b[n - 1] == 0):
                    print("He phuong trinh co vo so nghiem")
                else:
                    print("He phuong trinh vo nghiem")
                    
        # Khu cac hang ben duoi 
        for r in range(row + 1, n):
            if A[r][col] == 0:
                continue
            
            factor = Fraction(A[r][col], A[row][col])
            step += 1
            
            factor_str = format_frac(factor)
            print(f"\n Bước {step}: R{r+1} = R{r+1} - ({factor_str}) × R{row+1}")
            A[r][col] = factor
            for j in range(col + 1, n, 1):
                A[r][j] = A[r][j] - factor*A[row][j]
                
            print_matrix(A, n)
        
        row += 1
        
    return A, p

print(" A = PLU completed")

def swap_row(arr, index1, index2):
    if isinstance(arr, np.ndarray):
        arr[[index1, index2]] = arr[[index2, index1]]  # ✅ numpy-safe
    else:
        arr[index1], arr[index2] = arr[index2], arr[index1]  # list bình thường
```

### create matrix_explore

```python
def matrix_explore(A, b):
    arr = []
    n = len(A)
    
    for i in range(n):
        row = []
        for j in range(n):
            row.append(Fraction(A[i][j]))
        row.append(Fraction(b[i]))
        arr.append(row)
    
    print_matrix_explore(arr, n)
    return arr
```
