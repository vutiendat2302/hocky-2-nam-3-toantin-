## Phuong phap so gia dung he phuong trinh ax = b


- Phuong phap khu gauss
- Phuong phap phan tich LU

## So sanh 2 phuong phap:
- theo y hieu ca nhan thi khu gauss se bi phu thuoc vao b, neu ma tran a (nxn) => kho giai quyet
- phan tich LU tuy cung di khu gauss (khong doi cho cac hang) -> tim ma tran L, nhung he nay khong phu thuoc vao b
    - Thuc hien khu Gauss khong doi vi tri cac hang, cac cot -> ta se phan tich duoc A = LU voi U = ma tran A sau khi khu gauss, L la cac he so khi khu cac hang
    - Dieu kien van giong khu gauss la pivot != 0, neu pivot = 0 -> phuong phap khac la nhan them ma tran hoan vi 
    - ta se di tim L (ma tran tam giac duoi), U(ma tran tam giac tren) -> giai he Ly = b, tim duoc y giai Ux = y
    - khong phu thuoc vao b co dinh, neu co thay doi b thi van co the tai su dung 
    - so buoc thuc hien cua LU it hon Gauss (2 * n^2) con Gauss la (n^3 / 3)
    

## Phuong phap Gauss


```python
import numpy as np
import pandas as pd
from fractions import Fraction
```

```python
# print matrix 

def to_frac(matrix):
    """Chuyển ma trận sang dạng phân số để hiển thị chính xác"""
    rows = len(matrix)
    cols = len(matrix[0])
    return [[Fraction(matrix[i][j]).limit_denominator(1000) for j in range(cols)] for i in range(rows)]

def format_frac(f):
    """Format phân số đẹp"""
    if f.denominator == 1:
        return str(f.numerator)
    return f"{f.numerator}/{f.denominator}"

def print_matrix(matrix, n, step_title=""):
    """In ma trận bổ sung [A|b] đẹp"""
    if step_title:
        print(f"\n{'-'*60}")
        print(f"  {step_title}")
        print(f"{'-'*60}")
    
    col_width = 10
    
    # Header
    header = "    "
    for j in range(n):
        header += f"  x{j+1}".center(col_width)
    header += " | "
    header += "  b".center(col_width)
    print(header)
    print("    " + "-" * (col_width * (n + 1) + 3))
    
    for i in range(len(matrix)):
        row_str = f" R{i+1} |"
        for j in range(n):
            val = format_frac(matrix[i][j])
            row_str += str(val).center(col_width)
        row_str += " | "
        val_b = format_frac(matrix[i][n])
        row_str += str(val_b).center(col_width)
        print(row_str)
    
    print("    " + "-" * (col_width * (n + 1) + 3))

print("Print completed!")
```

```python
# Tạo ma trận mo rong
def matrix_explore(A, b, verbose = True):
    # arr = [[Fraction(A[i][j]) for j in range(n)] + [Fraction(b[i])] for i in range(n)]
    arr = []
    n = len(A)
    for i in range(n):
        row = []
        for j in range(n):
            row.append(Fraction(A[i][j]))
        
        row.append(Fraction(b[i]))
        arr.append(row)
        
    if verbose:
        print("Phuong phap Khu Gauss")
        print(f"Ma tran {n} x {n}")
        print_matrix(arr, n, "Ma trận mo rong [A|b]")
    
    return arr;
```

```python
# Phuong phap the 
def pp_the(arr, step, n, verbose = True):
    x = [0] * n
    for i in range(n - 1, -1, -1):
        if arr[i][i] == 0:
            continue
        
        # Chia hàng cho phần tử chéo
        if arr[i][i] != 1:
            step += 1
            divisor = arr[i][i]
            if verbose:
                print(f"\n Bước {step}: R{i+1} = R{i+1} ÷ {format_frac(divisor)}")
            for j in range(n + 1):
                arr[i][j] = arr[i][j] / divisor
            if verbose:
                print_matrix(arr, n, " ")
        
        # Khử các hàng phía trên
        for r in range(i - 1, -1, -1):
            if arr[r][i] == 0:
                continue
            factor = arr[r][i]
            step += 1
            if verbose:
                factor_str = format_frac(factor)
                print(f"\n Bước {step}: R{r+1} = R{r+1} - ({factor_str}) × R{i+1}")
            for j in range(n + 1):
                arr[r][j] = arr[r][j] - factor * arr[i][j]
            if verbose:
                print_matrix(arr, n, " ")
        
        x[i] = arr[i][n]
    
    if verbose:
        print("Key")
        for i in range(n):
            val = format_frac(x[i])
            float_val = float(x[i])
            print(f"  x{i+1} = {val:>10}  ≈  {float_val:.6f}")
    
    for i in range(n):
        print(f"x{i + 1} = {x[i]} \n")
```

```python
# Phep khu Gauss
def gauss(A, b, verbose=True):
    n = len(A)
    arr = matrix_explore(A, b, "");
    step = 0
    
    # Khu gauss    
    pivot_cols = []  # Lưu vị trí pivot
    row = 0
    
    for col in range(n):
        if row >= n:
            break
        
        # Tìm phần tử pivot (tìm hàng không bằng 0)
        pivot_row = None
        for r in range(row, n):
            if arr[r][col] != 0:
                pivot_row = r
                break
        
        if pivot_row is None:
            continue
        
        # Đổi hàng nếu cần
        if pivot_row != row:
            arr[row] = arr[pivot_row]
            arr[pivot_row] = arr[row]
            step += 1
            if verbose:
                print(f"\n Bước {step}: R{row+1} ↔ R{pivot_row+1} ")
                print_matrix(arr, n, "")
        
        pivot_val = arr[row][col]
        pivot_cols.append((row, col))
        
        # Khử các hàng bên dưới
        changed = False
        for r in range(row + 1, n):
            if arr[r][col] == 0:
                continue
            
            factor = arr[r][col] / pivot_val
            
            step += 1
            if verbose:
                factor_str = format_frac(factor)
                print(f"\n Bước {step}: R{r+1} = R{r+1} - ({factor_str}) × R{row+1}")
            
            for j in range(n + 1):
                arr[r][j] = arr[r][j] - factor * arr[row][j]
            
            changed = True
            
            if verbose:
                print_matrix(arr, n, "")
        
        row += 1
    
    rank_A = row
    rank_Ab = rank_A
    
    for r in range(rank_A, n):
        if arr[r][n] != 0:
            rank_Ab = r + 1
            break
    
    if verbose:
        print(f"  • Hạng ma trận A:    rank(A)  = {rank_A}")
        print(f"  • Hạng ma trận [A|b]: rank(Ab) = {rank_Ab}")

    
    # Vô nghiệm
    if rank_Ab > rank_A:
        if verbose:
            print(f"\n Vo nghiem")
            
            # Tìm hàng mâu thuẫn
            for r in range(n):
                all_zero = all(arr[r][j] == 0 for j in range(n))
                if all_zero and arr[r][n] != 0:
                    print(f"   → Hàng R{r+1}: 0 = {format_frac(arr[r][n])} (mâu thuẫn!)")
        return None
    
    # Vô số nghiệm
    if rank_A < n:
        if verbose:
            print(f"\n Vo so nghiem")
            print(f"   Vì rank(A) = rank(A|b) = {rank_A} < n = {n}")
            print(f"   Có {n - rank_A} ẩn tự do")
        return None
    
    # Có nghiệm duy nhất
    if verbose:
        print(f"\n Co nghiem duy nhat")

    pp_the(arr, step, n, "")

print(" Gauss Completed ")
```

## Phuong phap phan tich LU 
- Van se dung khu Gauss nhung khong doi vi tri cac hang neu pivot = 0 => loi

```python
# print matrix 

def to_frac(matrix):
    """Chuyển ma trận sang dạng phân số để hiển thị chính xác"""
    rows = len(matrix)
    cols = len(matrix[0])
    return [[Fraction(matrix[i][j]).limit_denominator(1000) for j in range(cols)] for i in range(rows)]

def format_frac(f):
    """Format phân số đẹp"""
    if f.denominator == 1:
        return str(f.numerator)
    return f"{f.numerator}/{f.denominator}"

def print_matrix_expolre(matrix, n):
    col_width = 10
    # Header
    header = "    "
    for j in range(n):
        header += f"  x{j+1}".center(col_width)
    header += " | "
    header += "  b".center(col_width)
    print(header)
    print("    " + "-" * (col_width * (n + 1) + 3))
    
    for i in range(len(matrix)):
        row_str = f" R{i+1} |"
        for j in range(n):
            val = format_frac(matrix[i][j])
            row_str += str(val).center(col_width)
        row_str += " | "
        val_b = format_frac(matrix[i][n])
        row_str += str(val_b).center(col_width)
        print(row_str)
    
    print("    " + "-" * (col_width * (n + 1) + 3))
    
    
def print_matrix(matrix, n):
    col_width = 10
    
    # Header
    header = "    "
    for j in range(n):
        header += f"x{j+1}".center(col_width)
    
    print(header)
    print("    " + "-" * (col_width * n))
    
    # Nội dung
    for i in range(n):
        row_str = f" R{i+1} |"
        for j in range(n):
            val = format_frac(matrix[i][j])
            row_str += str(val).center(col_width)
        
        print(row_str)
    
    print("    " + "-" * (col_width * n))
print("Print completed!")
```

```python
# Phep phan tich LU
def LU(A, b, step):
    n = len(A)
    L = [[Fraction(0)] * n for _ in range(n)]
    
    for i in range(n):
        L[i][i] = 1
    
    # Khu gauss ma tran A
    pivot_cols = []
    row = 0
    
    for col in range(n):
        if row >= n:
            break
        
        # Tìm phần tử pivot (tìm hàng không bằng 0)
        pivot_row = row
        
        if A[row][col] == 0:
            if (row != n - 1):
                print("Khong the phan tich A thanh ma tran LU")
            else:
                if (b[n - 1] == 0):
                    print("He phuong trinh co vo so nghiem")
                else:
                    print("He phuong trinh vo nghiem")
        
        pivot_val = A[row][col]
        
        # Khử các hàng bên dưới
        changed = False
        for r in range(row + 1, n):
            if A[r][col] == 0:
                continue
            
            factor = Fraction(A[r][col], pivot_val)
            
            step += 1

            factor_str = format_frac(factor)
            print(f"\n Bước {step}: R{r+1} = R{r+1} - ({factor_str}) × R{row+1}")
            
            L[r][col] = factor
            
            for j in range(n):
                A[r][j] = A[r][j] - factor * A[row][j]
            
            changed = True
            print_matrix(A, n)
        
        row += 1
    
    # Có nghiệm duy nhất
    print(f"\n Co nghiem duy nhat")
    print("\n Ma tran tam giac tren (U) la: ")
    print_matrix(A, n)
    print("\n Ma tran tam giac duoi (L) la: ")
    print_matrix(L, n)

    return A, L
    
print(" A = LU Completed ")
```

```python
# Phuong phap the 
def pp_the(arr, step, n):
    x = [0] * n
    for i in range(n - 1, -1, -1):
        if arr[i][i] == 0:
            continue
        
        # Chia hàng cho phần tử chéo
        if arr[i][i] != 1:
            step += 1
            divisor = arr[i][i]
            print(f"\n Bước {step}: R{i+1} = R{i+1} ÷ {format_frac(divisor)}")
            
            for j in range(n + 1):
                arr[i][j] = arr[i][j] / divisor
                
            print_matrix(arr, n)
        
        # Khử các hàng phía trên
        for r in range(i - 1, -1, -1):
            if arr[r][i] == 0:
                continue
            factor = arr[r][i]
            step += 1
            
            factor_str = format_frac(factor)
            print(f"\n Bước {step}: R{r+1} = R{r+1} - ({factor_str}) × R{i+1}")
            
            for j in range(n + 1):
                arr[r][j] = arr[r][j] - factor * arr[i][j]
            print_matrix(arr, n)
        
        x[i] = arr[i][n]
    
    print("Key")
    for i in range(n):
        val = format_frac(x[i])
        float_val = float(x[i])
        print(f"  x{i+1} = {val:>10}  ≈  {float_val:.6f}")
    
    return x
```

```python
# Tạo ma trận mo rong
def matrix_explore(A, b):
    # arr = [[Fraction(A[i][j]) for j in range(n)] + [Fraction(b[i])] for i in range(n)]
    arr = []
    n = len(A)
    for i in range(n):
        row = []
        for j in range(n):
            row.append(Fraction(A[i][j]))
        
        row.append(Fraction(b[i]))
        arr.append(row)

    print_matrix_expolre(arr, n)
    return arr;
```

```python
# Tim nghiem cua he phuong trinh:

def find_x_LU(A, b):
    n = len(A)
    step = 0
    U, L = LU(A, b, step)
    
    # Su dung pp the giai phuong trinh Ly = b
    print("Giai bai toan Ly = b")
    print("ma tran mo rong cung Ly = b")
    
    arr1 = matrix_explore(U, b)
    y = pp_the(arr1, step, n)
    
    print("Giai bai toan Ux = y")
    arr2 = matrix_explore(L, y)
    x = pp_the(arr2, step, n)
    
```
