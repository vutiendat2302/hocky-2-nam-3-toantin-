# Giai bai tap Newtorn

```python
import math
import sympy as sp 
```

```python

def check_phanly(f, df, dff, a, b):
    print("\n2. Kiểm tra khoảng phân ly")
    print("f(", a, ")", round(f(a),6))
    print("f(", b, ")", round(f(b),6))
    if f(a)*f(b) < 0:
        print("f(", a, ")", "*", "f(", b, ")<0")
        print("Phương trình f(x) = 0 co nghiem trong khoang (", a, "," , b, ")")
    else:
        print("f(", a, ")", "*", "f(", b, ")>0")
        print("Phương trình f(x) = 0 khong co nghiem trong khoang (", a, "," , b, ")")

    if df(a) > 0 & df(b) > 0:
        print("f'(x) > 0 voi moi x thuoc (", a, ",", b, ")")
        print("f(x) la day don dieu tang") 
        print("khoang (", a, "," , b, ") la khoang phan ly nghiem cua phuong trinh f(x) = 0")
    elif f(a) < 0 & df(b) < 0:
        print("f'(x) > 0 voi moi x thuoc (", a, ",", b, ")")
        print("f(x) la day don dieu giam ") 
        print("khoang (", a, "," , b, ") la khoang phan ly nghiem cua phuong trinh f(x) = 0")
    else:
        print("Dung bang bien thien")
```

```python
# check f', f'' khong doi dau tren doan a, b
def check_khongdoidau(df, dff, a, b):
    if df(a)*df(b) > 0: 
        print("f'(x) > 0 voi moi x trong khoang (", a, "," , b, ") nen f'(x) khong doi dau tren (", a, "," , b, ")")
    elif df(a)*df(b) < 0:
        print("f'(x) doi dau tren khoang (", a, "," , b, ")")

    if dff(a)*dff(b) > 0: 
        print("f''(x) > 0 voi moi x trong khoang (", a, "," , b, ") nen f'(x) khong doi dau tren (", a, "," , b, ")")
    elif dff(a)*dff(b) < 0:
        print("f''(x) doi dau tren khoang (", a, "," , b, ")")
```

```python
# Chon diem fourier
def chose_fourier(f, dff, a, b):
    if f(a)*dff(a) > 0:
        x_0 = a
        print("Chon x0 = ", a, ", do f(", a, ")*", "f''(", a, ") > 0 nen x0 la diem fourier cua ham so f(x)")
    elif f(b)*dff(b) > 0:
        x_0 = b
        print("Chon x0 = ", b, ", do f(", b, ")*", "f''(", b, ") > 0 nen x0 la diem fourier cua ham so f(x)")
    else:
        print("ham f(x) khong ton tai diem fourier tren khoang (", a, "," , b, ")")

    return x_0 
```

```python

def caculator_m_M(df, dff, a, b):
    print("\n3. Tính M và m")
    M = max(abs(dff(a)), abs(dff(b)))
    m = min(abs(df(a)), abs(df(b)))
    print("M = max|f''(x)| =", M)
    print("m = min|f'(x)| =", m)
    C = M/(2*m)
    print("C =", C)
    return m, M, C
```

```python
def newtorn_table(f, df, dff, m, M, C, x_0, a, b, eps):
    print("Ap dung cong thuc lap Newtorn")
    print("x_n = x_n-1 - ", f, "/", df)
    print("Voi sai so |x_n - x*| <= ", C , "* (x_n - x_n-1)^2")
    print("Ta co bang sau:")
    print("\nn      xn         f(xn)       ", C , "* (x_n - x_n-1)^2")
    
    n = 0
    
    while True:
        
        fx = f(x_0)
        dfx = df(x_0)
        
        xn1 = x_0 - fx/dfx
        
        if n == 0:
            error = ""
        else:
            error = C * abs(xn-x_prev)
        
        print(f"{n:<2}  {xn:<12.6f} {fx:<12.6f} {error:<12.6f}")
        
        if n>0 and C * abs(xn-x_prev) < eps:
            break
        
        x_prev = x_0
        xn = xn1
        n += 1
    
    print("\n5. Kết luận")
    print("Nghiệm gần đúng:", xn)

```

```python
# khai báo biến
x = sp.symbols('x')

# hàm bài toán
f = sp.log(x-1) + sp.cos(x-1)

# đạo hàm
df = sp.diff(f, x)
ddf = sp.diff(df, x)

# chuyển sang hàm số để tính
f_num = sp.lambdify(x, f, "math")
df_num = sp.lambdify(x, df, "math")
ddf_num = sp.lambdify(x, ddf, "math")

# khoảng
a = 1.3
b = 2
eps = 1e-5

print("1. Hàm số")
print("f(x) =", f)
print("f'(x) =", df)
print("f''(x) =", ddf)

check_phanly(f_num, df_num, dff_num, a, b)
check_khongdoidau(df_num, dff_num, a, b)

x_0 = chose_fourier(f, dff, a, b)
m, M, C = caculator_m_M(df, dff, a, b)
newtorn_table(f_num, df_num, dff_num, m, M, C, x_0, a, b, eps)
```
