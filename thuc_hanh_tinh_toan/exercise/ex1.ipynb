## Vi du 2: Xac dinh nghiem cua phuong trinh bac 2 voi 
- a = 1, b = 62.10, c = 3
- chi su dung 4 chu so co nghia (Day la dieu kien) vd: 62.10
- so sanh sai so tuong doi xap xi cua cong thuc truyen thong va cong thuc cai tien 

```python
from decimal import Decimal, getcontext
import numpy  as np

getcontext().prec = 4 # thiet lap 4 chu so co nghia

a = Decimal('1') # phai de o dang String khong thi python se convert sang kieu float 
b = Decimal('62.10') # gay ra sai lech
c = Decimal('3')

# co the su dung numpy de tinh sqrt nhung no se convert sang float
def sqrt_decimal(x):
    return x.sqrt()

# Cong thuc truyen thong 
x1_old = (-b + sqrt_decimal(b**2 - 4 * a *c)) / (2 * a)
x2_old = (-b - sqrt_decimal(b**2 - 4 * a *c)) / (2 * a)

# Cong thuc cai tien 
x1_mod = (2 * c) / (-b - sqrt_decimal(b**2 - 4*a*c))
x2_mod = (2 * c) / (-b + sqrt_decimal(b**2 - 4*a*c))

# tang do chinh xac de so sanh sai so, vi de bai 
# chua cho nghiem chinh xac la bao nhieu 
getcontext().prec = 10

# Tinh nghiem x chuan 
x1_good = (-Decimal('62.10') + sqrt_decimal(Decimal('62.10')**2 \
    -4 * Decimal('1') * Decimal('3'))) / (2 * Decimal('1'))

x2_good = (-Decimal('62.10') - sqrt_decimal(Decimal('62.10')**2 \
    -4 * Decimal('1') * Decimal('3'))) / (2 * Decimal('1'))

err1 = np.abs((x1_good - x1_old) / x1_good * 100)
err2 = np.abs((x1_good - x1_mod) / x1_good * 100)
print(f" Nghiem chinh xac: {x1_good}")
print(f"Cong thuc truyen thong: {x1_old} voi sai so xap xi la: {err1 : .2f}%")
print(f"Cong thuc cai tien: {x1_mod} voi sai so xap xi la: {err2 : .2f}%")
```

# 3.6 Hay tinh gan dung e^-5 bang 2 cach
## C1: Khi trien taylor truc tiep
- Nhuoc diem: voi sai so lon, thi 2 so co the triet tieu di nhau 
## C2: dung nghich dao 
### Yeu cau:
- Dung 20 so hang cho moi cach de tinh e^-5
- So sanh voi gia tri dung la e^-5 = 6.737947 x 10 -3 


```python
import math
from math import factorial
import numpy as np
```

```python
true_value = 6.737947e-3
x = 5
n = 20

print("=== Method1: Taylor ===")
approx = 0

for i in range(n):
    term = ((-1)**i) * (x**i) / factorial(i)
    approx_prev = approx 
    approx += term
    
    # sai so:
    # sai so tuong doi:
    true_error = abs((true_value - approx) / true_value)
    # Sai so xap xi
    if i > 0:
        approx_error = abs((approx - approx_prev) / approx) 
    else:
        approx_error = 0
        
    print(f"Term {i+1:2d}: approx = {approx:.10f}, "
        f"true_err = {true_error:.6f}, approx_err = {approx_error:.6f}")

    
print("\n===== METHOD 2: 1 / e^x =====")
approx_exp = 0

for i in range(n):
    term = (x**i) / factorial(i)
    prev_exp = approx_exp
    approx_exp += term
    
    approx = 1 / approx_exp
    
    # Sai số
    true_error = abs((true_value - approx) / true_value)
    approx_error = abs((approx_exp - prev_exp) / approx_exp) if i > 0 else 0
    
    print(f"Term {i+1:2d}: approx = {approx:.10f}, "
        f"true_err = {true_error:.6f}, approx_err = {approx_error:.6f}")
```

# 3.7: Xet ham f(x)
- Cho dao ham cua f(x)
### Cau hoi:
- Xem co gap kho khan khi tinh tai x = 0.577 khong. Tinh f'(x) tai x = 0.577
- hay thu tinh voi 3 chu so, 4 chu so
- Dung chopping cat so, khong duoc lam tron

### Nhan xet: 
- Do mau so rat nho, voi x = 0.577 se cho ket qua lon voi sai so lon, 

```python
import math

def chop(x, digits):
    if x == 0:
        return 0
    sign = -1 if x < 0 else 1
    x = abs(x)
    exp = int(math.floor(math.log10(x)))
    factor = 10**(digits - 1 -exp)
    return sign * math.floor(x * factor) /factor

def compute(x, digits):
    # bước 1: x^2
    x = chop(x, digits)
    x2 = chop(x * x, digits)
    
    # bước 2: 3x^2
    t1 = chop(3 * x2, digits)
    
    # bước 3: 1 - 3x^2
    denom = chop(1 - t1, digits)
    
    # bước 4: (1 - 3x^2)^2
    denom_sq = chop(denom * denom, digits)
    
    # bước 5: 6x
    num = chop(6 * x, digits)
    
    # bước 6: chia
    result = chop(num / denom_sq, digits)
    
    return result, denom
    
x = 0.577

print("True value:")
true_val = (6*x)/((1 - 3*x*x)**2)
print(true_val)

print("\n--- 3-digit chopping ---")
res3, denom3 = compute(x, 3)
print("denominator:", denom3)
print("result:", res3)

print("\n--- 4-digit chopping ---")
res4, denom4 = compute(x, 4)
print("denominator:", denom4)
print("result:", res4)
```

- 3 digits cho thay sai lech cuc lon, nguyen nhan do sai so binh phuong o duoi mau khech dai manh
- 4 digits  do lech it hon 3 chu so nhung van sai dang ke

# 3.8: tinh y = x^3 - 7x^2 + 8x - 0.35
- tai x = 1.37
- dung 3 digits 
- dung chopping
- Sau do tinh percent relative error: sai so tuong doi ( no the hien ket qua lech bao nhieu 
so voi gia tri that)
- b, Viet lai da thuc theo kieu y = ((x - 7)x + 8) - 0.35 roi lm nhu a


```python
def chop(x, digits=3):
    if x == 0:
        return 0
    sign = -1 if x < 0 else 1
    x = abs(x)
    exp = int(math.floor(math.log10(x)))
    factor = 10**(digits - 1 - exp)
    return sign * math.floor(x * factor) / factor

# Giá trị thật
x_true = 1.37
y_true = x_true**3 - 7*x_true**2 + 8*x_true - 0.35

# -------- (a) dạng thường --------
x = chop(1.37)

x2 = chop(x * x)
x3 = chop(x2 * x)

t1 = chop(7 * x2)
t2 = chop(8 * x)

y_a = chop(x3 - t1)
y_a = chop(y_a + t2)
y_a = chop(y_a - 0.35)

# -------- (b) Horner --------
x = chop(1.37)

t = chop(x - 7)
t = chop(t * x)
t = chop(t + 8)
t = chop(t * x)
y_b = chop(t - 0.35)

# -------- Sai số --------
err_a = abs((y_true - y_a) / y_true) * 100
err_b = abs((y_true - y_b) / y_true) * 100

print("True value:", y_true)

print("\n(a) Normal form:")
print("y =", y_a)
print("Percent relative error =", err_a)

print("\n(b) Horner form:")
print("y =", y_b)
print("Percent relative error =", err_b)
```

# 3.11: Su dung 5 chu so co nghia voi phuong phap chopping de tim nghiem cua phuong trinh sau
- Su dung cac cong thuc eqs
- x^2 - 5000.002x + 10 = 0

```python
def chop(x, digits):
    """Chopping to n significant digits"""
    if x == 0:
        return 0
    sign = -1 if x < 0 else 1
    x = abs(x)
    exp = int(math.floor(math.log10(x)))
    factor = 10**(digits - 1 - exp)
    return sign * math.floor(x * factor) / factor

# Hệ số phương trình
a = 1
b = -5000.002
c = 10
digits = 5  # 5-digit arithmetic

# Bước 1: Tính delta = b^2 - 4ac
b_chop = chop(b, digits)
b2 = chop(b_chop * b_chop, digits)
four_ac = chop(4 * chop(a * c, digits), digits)
delta = chop(b2 - four_ac, digits)
sqrt_delta = chop(math.sqrt(delta), digits)

# Bước 2: Tính nghiệm lớn bằng công thức bình thường (Eq 3.12)
x1 = chop((-b_chop + sqrt_delta) / chop(2 * a, digits), digits)

# Bước 3: Tính nghiệm nhỏ bằng công thức bù (Eq 3.13)
y1 = chop(2 * c / (-b_chop + sqrt_delta), digits)

print("5-digit chopping results:")
print("Nghiệm lớn x1 =", x1)
print("Nghiệm nhỏ x2 =", y1)
```
