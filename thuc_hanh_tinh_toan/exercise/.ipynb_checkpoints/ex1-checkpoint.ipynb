## Vi du 1: 
- Uoc luong sai so cho cac phuong phap lap. 
- Xuất phát từ phiên bản đơn giản nhất, ex = 1, bằng cách cộng thêm dần các số hạng hãy tính xấp xỉ e0.5. 
- Tìm sai số tương đối thực và sai số tương đối xấp xỉ tại mỗi bước. Cần bao nhiêu số hạng để ước lượng cho sai số xấp xỉ εa không vượt quá dung sai εs phù hợp với 3 chữ số có nghĩa (cho e0.5 = 1.648721)

```python
# import library 
import math

# phan tich de bai thi ta co the tinh duoc dung sai bang cong thuc scarborough 
# esilon(s) = (0,5 x 10^(2-n))%
```

```python
# Giá trị thật
true_value = 1.648721
x = 0.5

# Sai số cho 3 chữ số có nghĩa (dung sai) 
es = 0.5 * 10**(2 - 3)  # = 0.05%

# Khởi tạo
n = 0 # dem so lan lap 
approx = 0 # 
prev_approx = 0
ea = 100  # sai số xấp xỉ ban đầu lớn
et = 0

print(f"{'n':<5}{'result':<15}{'epsilon_t (%)':<20}{'epsilon_a (%)':<20}")

while True:
    # Tính số hạng
    term = (x**n) / math.factorial(n)
    
    prev_approx = approx
    approx += term
    
    # Sai số thực
    et = abs((true_value - approx) / true_value) * 100
    
    # Sai số xấp xỉ
    if n == 0:
        ea = None
    else:
        ea = abs((approx - prev_approx) / approx) * 100
    
    # In kết quả
    print(f"{n:<5}{approx:<15.8f}{et:<20.8f}{ea if ea else '---':<20}")
    
    # Điều kiện dừng
    if ea is not None and ea < es:
        break
    
    n += 1

print("\n👉 Số số hạng cần dùng:", n + 1)
```

## Vi du 2: Xac dinh nghiem cua phuong trinh bac 2 voi 
- a = 1, b = 62.10, c = 3
- chi su dung 4 chu so co nghia (Day la dieu kien) vd: 62.10
- so sanh sai so tuong doi xap xi cua cong thuc truyen thong va cong thuc cai tien 

```python
from decimal import Decimal, getcontext
import numpy  as np

getcontext().prec = 4 # thiet lap 4 chu so co nghia

a = Decimal('1') # phai de o dang String khong thi python se convert sang kieu float 
b = Decimal('62.10') # gay ra sai lech
c = Decimal('3')

# co the su dung numpy de tinh sqrt nhung no se convert sang float
def sqrt_decimal(x):
    return x.sqrt()

# Cong thuc truyen thong 
x1_old = (-b + sqrt_decimal(b**2 - 4 * a *c)) / (2 * a)
x2_old = (-b - sqrt_decimal(b**2 - 4 * a *c)) / (2 * a)

# Cong thuc cai tien 
x1_mod = (2 * c) / (-b - sqrt_decimal(b**2 - 4*a*c))
x2_mod = (2 * c) / (-b + sqrt_decimal(b**2 - 4*a*c))

# tang do chinh xac de so sanh sai so, vi de bai 
# chua cho nghiem chinh xac la bao nhieu 
getcontext().prec = 10

# Tinh nghiem x chuan 
x1_good = (-Decimal('62.10') + sqrt_decimal(Decimal('62.10')**2 \
    -4 * Decimal('1') * Decimal('3'))) / (2 * Decimal('1'))

x2_good = (-Decimal('62.10') - sqrt_decimal(Decimal('62.10')**2 \
    -4 * Decimal('1') * Decimal('3'))) / (2 * Decimal('1'))

err1 = np.abs((x1_good - x1_old) / x1_good * 100)
err2 = np.abs((x1_good - x1_mod) / x1_good * 100)
print(f" Nghiem chinh xac: {x1_good}")
print(f"Cong thuc truyen thong: {x1_old} voi sai so xap xi la: {err1 : .2f}%")
print(f"Cong thuc truyen thong: {x1_mod} voi sai so xap xi la: {err2 : .2f}%")
```
