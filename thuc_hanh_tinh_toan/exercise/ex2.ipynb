## Bai thuc hanh 1:
- Sử dụng khai triển chuỗi Taylor tại x0 = π/4 với n từ 0 tới 6 để tính xấp xỉ f(x) = cos x tại x = π/3. Cho biết giá trị thực của f(π/3) là 0.5 hãy tính các sai số tương đối thực. Các kết quả biểu diễn dạng bảng như sau

```python
import math
import pandas as pd
from math import factorial

def f(x):
    return math.cos(x)

# Đạo hàm tuần hoàn của cos(x)
def f_derivative_str(n):
    cycle = ['cos(x)', '-sin(x)', '-cos(x)', 'sin(x)']
    return cycle[n % 4]

x0 = math.pi / 4
x = math.pi / 3
h = x - x0

f_true = f(x)
results = []

# Tinh xap xi taylor
for n in range(7):
    Tn = 0
    for k in range(n + 1):
        # Tính giá trị số cho xấp xỉ
        cycle_values = [math.cos(x0), -math.sin(x0), -math.cos(x0), math.sin(x0)]
        Tn += cycle_values[k % 4] * h**k / math.factorial(k)
    
    rel_error = abs(Tn - f_true) / abs(f_true) * 100
    
    results.append({
        'n': n,
        'f^(n)(x0)': f_derivative_str(n),
        'T_n(x)': round(Tn, 8),
        'Relative Error (%)': round(rel_error, 10)
    })

df = pd.DataFrame(results)
print(df)
```

# 4.2 Khai trien Maclaurin cua ham cos la:
- Bat dau tu phien ban don gian, lan luot them tung hang tu de xap xi cos(pi/3)
- Sau khi them hang tu moi
- Tinh loi phan tram thuc dua tren gia tri thuc tu may tinh
- tinh sai so sap si
- Tiep tuc them hang tu cho den khi gia tri tuyet doi cua loi phan tram xap xi nho hon tieu chuan loi duoc xac dinh voi 2 chu so co nghia


```python
import math

# xac dinh dung sai voi 2 chu so co nghia
x = math.pi / 3 
true_value = math.cos(x)

# dung sai 
error_criterion = (0.5*10**(2 - 2))


approx = 0
n = 0
prep_approx = 0
print(f"{'n':>2} | {'Term added':>15} | {'Approx':>12} | {'True rel % err':>15} | {'Approx rel % err':>18}")
print("-"*70)

while True:
    term = ((-1)**n) * (x**(2*n)) / math.factorial(2*n)
    approx += term
    true_err = abs((true_value - approx) / true_value) * 100

    if n == 0:
        approx_err = float('nan')
    else: 
        approx_err = abs((approx - prev_approx) / approx) * 100
    
    print(f"{n:2d} | {term:15.10f} | {approx:12.10f} | {true_err:15.10f} | {approx_err:18.10f}")
    
    # Kiểm tra điều kiện dừng (2 chữ số có nghĩa)
    if n > 0 and approx_err < error_criterion:
        break
    
    prev_approx = approx
    n += 1
```

# 4.3: Chuoi vo han xap xi e^x 
- a, Chung minh rang Macluarin la truong hop dac bien cua taylor 
- b, su dung taylor de xap xi f(x) = e^-x tai x(i + 1) = 1, xi = 0.2


```python
import math

# Hàm và các đạo hàm
def f(x):
    return math.exp(-x)

def f_derivative(x, n):
    """Trả về f^(n)(x) với f(x) = e^-x"""
    return (-1)**n * math.exp(-x)

# Điểm x_i và x_{i+1}
xi = 0.2
xi1 = 1.0
h = xi1 - xi

# Giá trị thực tại x_{i+1}
f_true = f(xi1)

print(f"Giá trị thực f({xi1}) = {f_true:.6f}\n")

# Tính xấp xỉ với bậc 0 -> 3
for n in range(4):
    Tn = 0
    for k in range(n+1):
        Tn += f_derivative(xi, k) * h**k / math.factorial(k)
    epsilon = abs(Tn - f_true)  # lỗi tuyệt đối
    print(f"Bậc {n}: Tn = {Tn:.6f}, |εt| = {epsilon:.6f}")
```

# 4.4: Su dung khai trien taylor bac 0 den bac 4 de du doan f(2.5) voi f(x) = lnx
- x0 = 1
- tinh sai so phan tram tuong doi cho moi xap xi 


```python
import math

# Hàm gốc
def f(x):
    return math.log(x)

# Đạo hàm bậc n tại x0
def f_derivative(n, x0):
    if n == 0:
        return math.log(x0)
    elif n == 1:
        return 1 / x0
    elif n == 2:
        return -1 / x0**2
    elif n == 3:
        return 2 / x0**3
    elif n == 4:
        return -6 / x0**4
    else:
        return 0  # chỉ tính tới bậc 4

x0 = 1
x = 2.5
h = x - x0
f_true = f(x)

print(f"Giá trị thực f({x}) = {f_true:.6f}\n")

for n in range(5):
    Tn = 0
    for k in range(n+1):
        Tn += f_derivative(k, x0) * h**k / math.factorial(k)
    rel_error = abs(Tn - f_true) / abs(f_true) * 100
    print(f"Taylor bậc {n}: T{n} = {Tn:.6f}, sai số % εt = {rel_error:.2f}%")
```

# 4.5 Su dung khai trien taylor bac 0 den bac 3 de du doan f(3)
- f(x)=25x3−6x2+7x−88
- x0 = 1, tinh sai so tuong doi cho moi xap xi 


```python
import math

# Hàm gốc
def f(x):
    return 25*x**3 - 6*x**2 + 7*x - 88

# Đạo hàm bậc n tại x0
def f_derivative(n, x0):
    if n == 0:
        return f(x0)
    elif n == 1:
        return 75*x0**2 - 12*x0 + 7
    elif n == 2:
        return 150*x0 - 12
    elif n == 3:
        return 150
    else:
        return 0  # chỉ tính tới bậc 3

x0 = 1
x = 3
h = x - x0
f_true = f(x)

print(f"Giá trị thực f({x}) = {f_true:.6f}\n")

for n in range(4):
    Tn = 0
    for k in range(n+1):
        Tn += f_derivative(k, x0) * h**k / math.factorial(k)
    rel_error = abs(Tn - f_true) / abs(f_true) * 100
    print(f"Taylor bậc {n}: T{n} = {Tn:.6f}, sai số % εt = {rel_error:.2f}%")
```
