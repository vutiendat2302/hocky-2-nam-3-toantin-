```python
# check dieu kien hoi tu
# check phi xem co phai la mot tu anh hay khong
def tu_anh(phi, dphi, a, b):
    if dphi(a) > 0 and dphi(b) > 0:
        print("phi'(x) > 0 voi moi x thuoc (", a, ",", b, ")")
        print("phi(x) la day don dieu tang") 
        min = phi(a)
        max = phi(b)
        print("min phi:", min)
        print("max phi", max)
        if (min >= a and max <= b):
            print("do phi(x) thuoc [", a, ",", b, "] voi moi x thuoc [", a, ",", b, "] nen phi(x) la mot",
                " tu anh tren [", a, ",", b, "]")
    elif dphi(a) < 0 and dphi(b) < 0:
        print("phi'(x) > 0 voi moi x thuoc (", a, ",", b, ")")
        print("phi(x) la day don dieu giam ") 
        min = phi(b)
        max = phi(a)
        print("min phi:", min)
        print("max phi", max)
        if (min >= a and max <= b):
            print("do phi(x) thuoc [", a, ",", b, "] voi moi x thuoc [", a, ",", b, "] nen phi(x) la mot",
                " tu anh tren [", a, ",", b, "]")
    else:
        print("khong ton tai tu anh")
```

```python
# check anh xa co 
def anh_xa_co(dphi, a, b):
    print("\n1. Kiểm tra điều kiện hội tụ")
    
    if (np.abs(dphi(a)) >= np.abs(dphi(b))):
        max = np.abs(dphi(a))
    else:
        max = np.abs(dphi(b))
        
    if max < 1:
        print("=> |phi'(x)| <= ", max,"< 1: Phương pháp lặp đơn hội tụ.",
            "phi(x) la mot anh xa co tren doan")
    else:
        print("=> Cảnh báo: Điều kiện hội tụ có thể không thỏa mãn tại điểm này.")
    
```

```python
# phuong phap lap don 
def lap_don(phi, x0, eps):
    print("\n2. Tiến hành lặp và lập bảng dữ liệu")
    print(f"{'n':<5} {'x_n':<15} {'x_{n+1}':<15} {'eps_a (%)':<15}")
    print("-" * 55)
    n = 0
    xn = x0
    
    while True:
        xn_next = phi(xn)
        
        # Tính sai số tương đối xấp xỉ eps_a
        if n == 0:
            eps_a = float('inf')
            eps_a_str = "---"
        else:
            eps_a = abs((xn_next - xn) / xn_next) * 100
            eps_a_str = f"{eps_a:.6f}%"
        
        # In dòng dữ liệu
        print(f"{n:<5} {xn:<15.6f} {xn_next:<15.6f} {eps_a_str:<15}")
        
        # Kiểm tra điều kiện dừng: eps_a <= 0.05%
        if n > 0 and eps_a <= eps:
            break
            
        xn = xn_next
        n += 1
        
        # Chặn vòng lặp vô hạn nếu có lỗi
        if n > 100: break

    print("-" * 55)
    print(f"3. Kết luận: Nghiệm xấp xỉ là x = {xn_next:.6f}")
    print(f"Số bước lặp: {n}")
```

```python
# khai báo biến
x = sp.symbols('x')

# hàm bài toán 
f = x - sp.exp(-x)
# chuyen ve dang x = phi(x)
phi = sp.exp(-x)
# đạo hàm
df = sp.diff(f, x)
ddf = sp.diff(df, x)

dphi = sp.diff(phi, x)
ddphi = sp.diff(dphi, x)

# chuyển sang hàm số để tính
f_num = sp.lambdify(x, f, "numpy")
df_num = sp.lambdify(x, df, "numpy")
ddf_num = sp.lambdify(x, ddf, "numpy")

phi_num = sp.lambdify(x, phi, "numpy")
dphi_num = sp.lambdify(x, dphi, "numpy")
ddphi_num = sp.lambdify(x, ddphi, "numpy")

# khoảng
a = 0
b = 2
eps = 0.05

print("1. Hàm số")
print("f(x) =", f)
print("f'(x) =", df)
print("f''(x) =", ddf)

print("phi(x) =", phi)
print("phi'(x) =", dphi)
print("phi''(x) =", ddphi)
 
check_phanly(f_num, df_num, a, b )
tu_anh(phi_num, dphi_num, a, b)
```

```python
# check f', f'' khong doi dau tren doan a, b
def check_khongdoidau(df, dff, a, b):
    if df(a)*df(b) > 0: 
        print("f'(x) > 0 voi moi x trong khoang (", a, "," , b, ") nen f'(x) khong doi dau tren (", a, "," , b, ")")
    elif df(a)*df(b) < 0:
        print("f'(x) doi dau tren khoang (", a, "," , b, ")")

    if dff(a)*dff(b) > 0: 
        print("f''(x) > 0 voi moi x trong khoang (", a, "," , b, ") nen f'(x) khong doi dau tren (", a, "," , b, ")")
    elif dff(a)*dff(b) < 0:
        print("f''(x) doi dau tren khoang (", a, "," , b, ")")
```

```python
# Chon diem fourier
def chose_fourier(f, dff, a, b):
    if f(a)*dff(a) > 0:
        x_0 = a
        print("Chon x0 = ", a, ", do f(", a, ")*", "f''(", a, ") > 0 nen x0 la diem fourier cua ham so f(x)")
    elif f(b)*dff(b) > 0:
        x_0 = b
        print("Chon x0 = ", b, ", do f(", b, ")*", "f''(", b, ") > 0 nen x0 la diem fourier cua ham so f(x)")
    else:
        print("ham f(x) khong ton tai diem fourier tren khoang (", a, "," , b, ")")

    return x_0 
```

```python

def caculator_m_M(df, dff, a, b):
    print("\n3. Tính M và m")
    M = max(abs(dff(a)), abs(dff(b)))
    m = min(abs(df(a)), abs(df(b)))
    print("M = max|f''(x)| =", M)
    print("m = min|f'(x)| =", m)
    C = M/(2*m)
    print("C =", C)
    return m, M, C
```

```python
import numpy as np
import sympy as sp

def check_phanly(f, a, b):
    print("\n1. Kiểm tra khoảng phân ly nghiệm")
    fa, fb = f(a), f(b)
    print(f"f({a}) = {fa:.6f}")
    print(f"f({b}) = {fb:.6f}")
    if fa * fb < 0:
        print(f"-> f({a}) * f({b}) < 0: Khoảng [{a}, {b}] chứa ít nhất một nghiệm.")
        return True
    else:
        print("-> Cảnh báo: Khoảng này không thỏa mãn điều kiện đổi dấu.")
        return False

def secant_table(f, x0, x1, eps):
    print("\n2. Tiến hành lặp Cát tuyến và lập bảng")
    print(f"{'n':<5} {'x_{n-1}':<12} {'x_n':<12} {'x_{n+1}':<12} {'f(x_{n+1})':<12} {'eps_a (%)':<12}")
    print("-" * 80)
    
    n = 1
    # Phương pháp cát tuyến cần 2 điểm khởi đầu
    xn_minus_1 = x0
    xn = x1
    
    while True:
        fxn = f(xn)
        fxn_minus_1 = f(xn_minus_1)
        
        # Kiểm tra tránh chia cho 0
        if fxn - fxn_minus_1 == 0:
            print("Lỗi: Mẫu số bằng 0.")
            break
            
        # Công thức Cát tuyến
        xn_plus_1 = xn - fxn * (xn - xn_minus_1) / (fxn - fxn_minus_1)
        fx_new = f(xn_plus_1)
        
        # Tính sai số tương đối xấp xỉ
        eps_a = abs((xn_plus_1 - xn) / xn_plus_1) * 100
        eps_a_str = f"{eps_a:.6f}%"
        
        # In dòng dữ liệu
        print(f"{n:<5} {xn_minus_1:<12.6f} {xn:<12.6f} {xn_plus_1:<12.6f} {fx_new:<12.6f} {eps_a_str:<12}")
        
        # Điều kiện dừng
        if eps_a <= eps:
            break
            
        # Cập nhật giá trị cho lần lặp kế tiếp
        xn_minus_1 = xn
        xn = xn_plus_1
        n += 1
        
        if n > 50: break

    print("-" * 80)
    print(f"3. Kết luận: Nghiệm xấp xỉ là x = {xn_plus_1:.6f} sau {n} bước lặp.")

# --- THỰC THI ---
x = sp.symbols('x')
f_sym = x - sp.exp(-x)
f_num = sp.lambdify(x, f_sym, "numpy")

# Thông số đầu vào
a, b = 0, 1      # Khoảng phân ly
x0, x1 = 0, 1    # Hai điểm xuất phát ban đầu
target_eps = 0.05 # 0.05%

if check_phanly(f_num, a, b):
    secant_table(f_num, x0, x1, target_eps)
```
