# Đánh giá mô hình 

```python
## Accuracy
## Trong python use scikit learning library caculater accuracy
## Gia su y_label la nhan dung cua du lieu và y_pred la nhãn dự báo theo 
## mô hình
## !pip install scikit-learn 
y_true = [0,1,1,1,0] 
y_pred = [1,1,1,1,0]

from sklearn.metrics import accuracy_score
accuracy_score(y_true, y_pred)
```

```python
## Precision 
## Tính Precision-recall dựa trên ground truth y_label và xác suất dự báo
## y_prob
## threshold là ngưỡng phân loại, chuyển các y_prob về nhãn, ví dụ >= 
## threshold thì -> predict = 1

y_label = [0,1,1,1,0] 
y_prob = [0.5,0.3 ,0.9,0.5, 0]
from sklearn.metrics import precision_recall_curve 
prec, rec, thres = precision_recall_curve(y_label, y_prob) 
print(prec, rec, thres)
```
