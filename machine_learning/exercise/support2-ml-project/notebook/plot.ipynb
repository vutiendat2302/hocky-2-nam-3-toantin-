```python
import matplotlib.pyplot as plt
import numpy as np

plt.rcParams['font.family'] = 'DejaVu Sans'
plt.rcParams['axes.spines.top'] = False
plt.rcParams['axes.spines.right'] = False

configs = ['Train 80%', 'Train 70%', 'Train 60%']
x = np.arange(len(configs))
w = 0.13

# MLP (không PCA)
mlp_r2_train   = [0.8043, 0.7991, 0.7913]
mlp_r2_val     = [0.7890, 0.7771, 0.7604]
mlp_rmse_train = [0.7510, 0.7558, 0.7627]
mlp_rmse_val   = [0.7628, 0.7740, 0.7893]
mlp_mae_val    = [0.4495, 0.4627, 0.4785]

# MLP + PCA
pca_r2_train   = [0.7483, 0.7422, 0.7352]
pca_r2_val     = [0.7417, 0.7345, 0.7219]
pca_rmse_train = [0.7997, 0.8045, 0.8094]
pca_rmse_val   = [0.8024, 0.8087, 0.8192]
pca_mae_val    = [0.5028, 0.5106, 0.5258]

C1 = '#1D4ED8'   # MLP R2
C2 = '#93C5FD'   # PCA R2
C3 = '#D97706'   # MLP RMSE
C4 = '#FCD34D'   # PCA RMSE
C5 = '#059669'   # MLP MAE
C6 = '#6EE7B7'   # PCA MAE

fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(17, 6))
fig.patch.set_facecolor('#FAFAFA')
fig.suptitle('MLP (α=100) — So sánh hiệu suất & Overfitting: MLP vs MLP+PCA',
             fontsize=13, fontweight='bold', color='#111827', y=1.02)

# ═══════════════════════════════════════════════
# BIỂU ĐỒ 1: So sánh R² Val, RMSE Val, MAE Val
# ═══════════════════════════════════════════════
ax1.set_facecolor('#F3F4F6')

offsets = [-2.5*w, -1.5*w, -0.5*w, 0.5*w, 1.5*w, 2.5*w]
bars_data = [
    (mlp_r2_val,   C1, 'R² Val (MLP)'),
    (pca_r2_val,   C2, 'R² Val (PCA)'),
    (mlp_rmse_val, C3, 'RMSE Val (MLP)'),
    (pca_rmse_val, C4, 'RMSE Val (PCA)'),
    (mlp_mae_val,  C5, 'MAE Val (MLP)'),
    (pca_mae_val,  C6, 'MAE Val (PCA)'),
]

for (data, color, label), off in zip(bars_data, offsets):
    bars = ax1.bar(x + off, data, w, label=label, color=color, alpha=0.9, edgecolor='white')
    for bar, v in zip(bars, data):
        ax1.text(bar.get_x() + bar.get_width()/2, v + 0.004,
                 f'{v:.3f}', ha='center', va='bottom', fontsize=6.2, rotation=90)

ax1.set_title('So sánh R² / RMSE / MAE (Validation)', fontweight='bold', color='#1F2937')
ax1.set_xticks(x); ax1.set_xticklabels(configs, fontsize=9)
ax1.set_ylabel('Giá trị'); ax1.set_ylim(0, 1.02)
ax1.legend(fontsize=7.5, ncol=2, loc='upper right')
ax1.grid(axis='y', alpha=0.35)

# ═══════════════════════════════════════════════
# BIỂU ĐỒ 2: Overfitting Train vs Val
# ═══════════════════════════════════════════════
ax2.set_facecolor('#F3F4F6')

# R² lines (trục trái)
ax2.plot(configs, mlp_r2_train, 'o--', color=C1, label='R² Train (MLP)', linewidth=2, markersize=7)
ax2.plot(configs, mlp_r2_val,   'o-',  color=C1, label='R² Val (MLP)',   linewidth=2, markersize=7, alpha=0.55)
ax2.fill_between(range(3), mlp_r2_train, mlp_r2_val, alpha=0.08, color=C1)

ax2.plot(configs, pca_r2_train, 's--', color=C5, label='R² Train (PCA)', linewidth=2, markersize=7)
ax2.plot(configs, pca_r2_val,   's-',  color=C5, label='R² Val (PCA)',   linewidth=2, markersize=7, alpha=0.55)
ax2.fill_between(range(3), pca_r2_train, pca_r2_val, alpha=0.08, color=C5)

# RMSE lines (trục phải)
ax2r = ax2.twinx()
ax2r.plot(configs, mlp_rmse_train, '^--', color=C3, label='RMSE Train (MLP)', linewidth=2, markersize=7)
ax2r.plot(configs, mlp_rmse_val,   '^-',  color=C3, label='RMSE Val (MLP)',   linewidth=2, markersize=7, alpha=0.55)
ax2r.fill_between(range(3), mlp_rmse_train, mlp_rmse_val, alpha=0.08, color=C3)

ax2r.plot(configs, pca_rmse_train, 'D--', color=C4, label='RMSE Train (PCA)', linewidth=2, markersize=7)
ax2r.plot(configs, pca_rmse_val,   'D-',  color=C4, label='RMSE Val (PCA)',   linewidth=2, markersize=7, alpha=0.55)
ax2r.fill_between(range(3), pca_rmse_train, pca_rmse_val, alpha=0.08, color=C4)

ax2r.set_ylabel('RMSE', color=C3, fontsize=10)
ax2r.tick_params(axis='y', labelcolor=C3)
ax2r.set_ylim(0.720, 0.850)
ax2r.spines['top'].set_visible(False)

ax2.set_title('Overfitting: Train vs Validation (R² & RMSE)', fontweight='bold', color='#1F2937')
ax2.set_ylabel('R²', color=C1, fontsize=10)
ax2.tick_params(axis='y', labelcolor=C1)
ax2.set_ylim(0.700, 0.840)
ax2.grid(axis='y', alpha=0.3)

# Gộp legend
l1, lb1 = ax2.get_legend_handles_labels()
l2, lb2 = ax2r.get_legend_handles_labels()
ax2.legend(l1 + l2, lb1 + lb2, fontsize=7, ncol=2, loc='lower left')

# Annotation delta
for i, (tr, vl) in enumerate(zip(mlp_r2_train, mlp_r2_val)):
    mid = (tr + vl) / 2
    ax2.annotate(f'Δ={tr-vl:+.4f}', xy=(i, mid), fontsize=6.5,
                 color=C1, ha='left', va='center',
                 xytext=(i+0.08, mid))

plt.tight_layout()
plt.show()
print("Done")
```
